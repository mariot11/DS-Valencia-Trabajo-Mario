## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

````python
def dia_de_la_semana(numero:int):
    lista_dias = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"]
    return lista_dias[numero - 1]

for i in range(1,8):
    print(dia_de_la_semana(i))
````

## Ejercicio 2
En el ejercicio 8 de bucles, creábamos una pirámide invertida, cuyo número de pisos venía determinado por un input del usuario. Crear una función que replique el comportamiento de la pirámide, y utiliza un único parámetro de entrada de la función para determinar el número de filas de la pirámide, es decir, elimina la sentencia input.

````python
def piramide_invertida(rango:int):
    for i in range(rango,0,-1):
        for j in range(i, 0, -1):
            if j <= 1:
                print(j)
            else:
                print(j, end = " ")

piramide_invertida(5)
piramide_invertida(4)
piramide_invertida(3)
````

Escibe una función que compare dos números. La función tiene dos argumentos y hay tres salidas posibles: que sean iguales, que el primero se  mayor que el segundo, o que el segundo sea mayor que el primero

````python

def comparacion(valor1:int, valor2:int):
    if(valor1 > valor2):
        print("El primero es mayor")
    elif(valor1 < valor2):
        print("El segundo es mayor")
    else:
        print("Son iguales")


comparacion(10,5)
comparacion(2,4)
comparacion(3,3)
````

## Ejercicio 4
Escribe una función que sea un contador de letras. En el primer argumento tienes que introducir un texto, y el segundo que sea la letra a contar. La función tiene que devolver un entero con el número de veces que aparece esa letra, tanto mayuscula, como minúscula

````python
def contador_letras(texto:str, letra:str):
    contador = 0
    for i in texto.upper():
        if letra.upper() == i:
            contador += 1
    return contador

print(contador_letras("En un lugar de la Mancha de cuyo nombre no quiero acordarme)", "m"))
print(contador_letras("En un lugar de la Mancha de cuyo nombre no quiero acordarme)", "M"))
print(contador_letras("En un lugar de la Mancha de cuyo nombre no quiero acordarme)", "e"))
print(contador_letras("En un lugar de la Mancha de cuyo nombre no quiero acordarme)", "E"))
````

## Ejercicio 5
Escribe una función que tenga un único argumento, un string. La salida de la función tiene que ser un diccionario con el conteo de todas las letras de ese string.

````python
def conteo(cadena:str):
    diccionario_conteo = {}

    for letra in cadena:
        diccionario_conteo.update({letra:diccionario_conteo.get(letra, 0) + 1})

    return diccionario_conteo

print(conteo("Abra Kadabra"))
````

## Ejercicio 6
Escribir una función que añada o elimine elementos en una lista. La función necesita los siguientes argumentos:
* lista: la lista donde se añadirán o eliminarán los elementos
* comando: "add" o "remove"
* elemento: Por defecto es None.

````python
lista_prueba = [1,2,3,4]

def modificar_lista(lista:list, comando:str, elemento = None):
    if comando.lower() == "add":
        lista.append(elemento)
    elif comando.lower() == "remove":
        try:
            lista.remove(elemento)
        except:
            print(f"El elemento {elemento} no se puede eliminar porque no existe en la lista")

modificar_lista(lista_prueba, "add", 5)
print(lista_prueba)
modificar_lista(lista_prueba, "remove", 3)
print(lista_prueba)
modificar_lista(lista_prueba, "remove", 7)
print(lista_prueba)
modificar_lista(lista_prueba, "add")
print(lista_prueba)
modificar_lista(lista_prueba, "remove")
print(lista_prueba)
````

## Ejercicio 7
Crea una función que reciba un número arbitrario de palabras, y devuelva una frase completa, separando las palabras con espacios.

````python

def completar_frase(*palabras:str):
    frase = ""
    for palabra in palabras:
        frase += str(palabra) + " "
    return frase[:-1]

print(completar_frase("Érase", "una", "vez", "en", "un", "lugar", "muy", "lejano"))
````

## Ejercicio 8
Escribe un programa que obtenga el enésimo número de la [serie de Fibonacci](https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci). Tienes que crear una función recursiva con un único argumento.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

````python
def fibonacci(n:int):
    if n < 3:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(2))
print(fibonacci(4))
print(fibonacci(5))
print(fibonacci(8))
````

## Ejercicio 9
Define en una única celda las siguientes funciones:
* Función que calcule el área de un cuadrado
* Función que calcule el area de un triángulo
* Función que calcule el área de un círculo

En otra celda, calcular el area de:
* Dos círculos de radio 10 + un triángulo de base 3 y altura 7
* Un cuadrado de lado = 10 + 3 círculos (uno de radio = 4 y los otros dos de radio = 6) + 5 triángulos de base = 2 + altura = 4

````python
def area_cuadrado(lado:float):
    return lado **2

def area_triangulo(base:float, altura:float):
    return base * altura / 2

def area_circulo(radio:float):
    import math
    return math.pi * (radio**2)
````

````python
print(area_circulo(10) * 2 + area_triangulo(3,7))

print(area_cuadrado(10) + area_circulo(4) + (area_circulo(6) * 2) + (area_triangulo(2, 4) * 5))
````

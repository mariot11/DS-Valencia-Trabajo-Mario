## Ejercicio 11
#### Bonus track. Mini hundir la flota
1. Crea un tablero de 10x10 relleno del carácter espacio ( ).
2. Posiciona un par de barcos en [(0,1), (1,1)] y [(1,3), (1,4), (1,5), (1,6)]. Los barcos serán Os mayúsculas. Como ves, un barco de dos posiciones de eslora y otro de cuatro.
3. Recibe un disparo en uno de los barcos, sustituyendo la O por una X
4. Recibe un disparo en agua, sustituyendo uno de los espacios por un guión: -
5. Ignorando los disparos recibidos, ahora prueba a posicionar otro barco de 4 posiciones dentro del tablero, pero **de manera ALEATORIA.**

**Consejo apartado 5**: El barco tendrá una posición inicial aleatoria dentro del tablero, y una orientación también aleatoria (N,S,E,O).

Por ejemplo, si tu posicion inicial fuese (5,0) y la orientación O, no podrias ponerlo porque se sale del tablero. Ahora bien, si fuese orientación este, si podrías, y ocuparía las posiciones [(5,0), (5,1), (5,2), (5,3)]. Por tanto tendrás que iterar con un bucle si es posible encajar el barco en el tablero, y el bucle terminará cuando lo consigas encajar. **Ojo con las posiciones donde ya hay barco!!!**

Usa `import random` para elegir las coordenadas aleatorias.

````python
import numpy as np

# 1.
tablero = np.full((10, 10), ' ')
print(tablero, end="\n\n")

# 2.
tablero[0:2, 1] = 'O'
tablero[1, 3:7] = 'O'

def recibir_disparo(fila:int, colum:int):
    if tablero[fila, colum] == " ":
        tablero[fila, colum] = "-"
    elif tablero[fila, colum] == "O":
        tablero[fila, colum] = "X"
    print(tablero, end="\n\n")

# 3.
recibir_disparo(4,5)
# 4.
recibir_disparo(1,3)

def posicionar_barco():
    colocado = False
    while not colocado:
        i = np.random.randint(10)
        j = np.random.randint(10)

        while (tablero[i,j] != " "):
            i = np.random.randint(10)
            j = np.random.randint(10)

        orientaciones = ["N","S","E","O"]

        while (len(orientaciones) > 0):
            orientacion = np.random.choice(orientaciones)
            match orientacion:
                case "N":
                    if (i >= 3):
                        if("O" not in tablero[i-3 : i+1, j]):
                            tablero[i-3 : i+1, j] = "O"
                            colocado = True
                            break
                case "S":
                    if (i <= 6):
                        if("O" not in tablero[i:i+4,j]):
                            tablero[i:i+4,j] = "O"
                            colocado = True
                            break
                case "E":
                    if (j <= 6):
                        if("O" not in tablero[i,j:j+4]):
                            tablero[i,j:j+4] = "O"
                            colocado = True
                            break
                case "O":
                    if (j >= 3):
                        if("O" not in tablero[i, j-3 : j+1]):
                            tablero[i, j-3 : j+1] = "O"
                            colocado = True
                            break
            
            orientaciones.remove(orientacion)

posicionar_barco()

print(tablero, end="\n\n")
````
